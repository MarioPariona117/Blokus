```python
# from copy import deepcopy

done = False
def encode_board(board):
    return ''.join(str(cell) for row in board for cell in row)

# def encode_board_alt(board):

def decode_board(encoded_board, n):
    board = []
    for i in range(n):
        row = [int(encoded_board[i * n + j]) for j in range(n)]
        board.append(row)
    return board

def compute_minimax(board_size, num_players, max_depth): ## for each state, I want the optimal action
    env = gym.make('gymnasium_env/Blokus-v0', board_size=board_size, num_players=num_players, render_mode='console', render_scale=10, disable_env_checker=True, mode = "good")
    env = env.unwrapped
    env.order_enforce = False
    best_action, minimax_value = {}, {}
    global different_states, function_calls
    different_states, function_calls = 0, 0
    
    def minimax(state:BlokusEnv, depth, obs, terminated=False) -> int:
        global function_calls, different_states
        function_calls += 1

        board_encoding = encode_board(state.board)
        if board_encoding in best_action:
            return best_action[board_encoding]
        different_states += 1
        if depth == 0 or terminated:
            return 0

        ret = float('-inf')
        possible_actions = obs["possible_actions"]
        if len(possible_actions) == 0:
            return 0
        hey = state.capture_state()
        if depth == max_depth:
            possible_actions = tqdm(possible_actions)
        for action in possible_actions:
            new_state = state
            new_obs, reward, terminated, truncated, info = new_state.step(action)
            if truncated:
                assert False
            eval = minimax_value[encode_board(new_state.board)] = minimax(new_state, depth - 1, new_obs, terminated)
            if ret < -eval + reward:
                best_action[board_encoding] = action
                ret = -eval + reward
                if depth == 100:
                    print(ret, action)
            state.restore_state(hey)
        return ret
    obs1, info = env.reset()
    minimax_value[encode_board(env.board)] = minimax(env, max_depth, obs1, False)

    return best_action, minimax_value, different_states, function_calls
    
```

```python
BOARD_SIZE = 5
best_action, minimax_value, different_states, function_calls = compute_minimax(board_size=BOARD_SIZE, num_players=2, max_depth=100)
print("different states", different_states)
print("function calls", function_calls)
print(minimax_value, different_states, function_calls)
print(env._decode_action(best_action["0" * BOARD_SIZE * BOARD_SIZE]), minimax_value["0" * BOARD_SIZE * BOARD_SIZE])
print(env._decode_action(best_action["0" * BOARD_SIZE * BOARD_SIZE]), minimax_value["0" * BOARD_SIZE * BOARD_SIZE])
```
