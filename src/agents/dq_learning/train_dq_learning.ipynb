```python
import numpy as np
import matplotlib.pyplot as plt

# Define a continuous loss function
def continuous_loss(predicted_values, target_values, threshold=-50):
    return np.where(predicted_values <= threshold, 0, (predicted_values - target_values))

# Example usage with continuous loss
predicted_values = np.linspace(-100, 100, 100)  # Predicted values from -100 to 100

# Compute loss
loss_values = continuous_loss(predicted_values, predicted_values)

# Plot the loss function
plt.plot(predicted_values, loss_values, label="Loss Function")
plt.axvline(x=-50, color='r', linestyle='--', label="Threshold (-50)")
plt.xlabel("Predicted Value")
plt.ylabel("Loss")
plt.title("Continuous Loss Function")
plt.legend()
plt.grid(True)
plt.show()
```
